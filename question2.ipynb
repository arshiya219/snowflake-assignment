## Remaining Objects: Warehouse & Stage

```
%%sql -r create_wh
CREATE WAREHOUSE IF NOT EXISTS ASSIGNMENT_WH
    WAREHOUSE_SIZE = 'XSMALL'
    AUTO_SUSPEND = 60
    AUTO_RESUME = TRUE;
```

```
%%sql -r create_stage
CREATE OR REPLACE STAGE ASSIGNMENT_STAGE
    COMMENT = 'Stage for assignment data files';
```

```
%%sql -r stages
SHOW STAGES;
```

## DML Operations: INSERT, UPDATE, DELETE

```
%%sql -r insert_more
INSERT INTO STUDENTS (STUDENT_ID, STUDENT_NAME, DEPARTMENT, MARKS) VALUES
(11, 'Karen Thompson', 'Biology', 90),
(12, 'Leo Martinez', 'Biology', 74);
```

```
%%sql -r update_result
UPDATE STUDENTS
SET MARKS = 78
WHERE STUDENT_ID = 4;
```

```
%%sql -r verify_update
SELECT * FROM STUDENTS WHERE STUDENT_ID = 4;
```

```
%%sql -r delete_result
DELETE FROM STUDENTS
WHERE STUDENT_ID = 12;
```

```
%%sql -r final_records
SELECT * FROM STUDENTS ORDER BY STUDENT_ID;
```
